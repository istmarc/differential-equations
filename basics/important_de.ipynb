# Important differential equations

```python
import numpy as np
import matplotlib.pyplot as plt
import scienceplots
plt.style.use("science")
import math
```

The solution of 

$$\dot{y} = ay$$

is

$$y = C e^{at}$$

where $C$ is a constant.

## Solution of $\dot{y} = ay$

```python
def f(a, c, t):
    n = t.size
    y = np.zeros(n)
    for i in range(n):
        y[i] = c*math.exp(a*t[i])
    return y
```

```python
# Plot solutions for c in [-1, 1.] and a = .5
m = 21
c = np.linspace(-1., 1., m)
n = 10
t = np.linspace(-1., 1., n)
y = np.zeros((n, m))
for j in range(m):
    y[:, j] = f(.5, c[j], t)
```

```python
for j in range(m):
    plt.plot(t, y[:, j])
plt.title(r"$y=C e^{at}$")
plt.xlabel(r"$x$")
plt.ylabel(r"$y$")
plt.show()
```

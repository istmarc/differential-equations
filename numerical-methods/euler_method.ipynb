# Euler method

```python
import numpy as np
import matplotlib.pyplot as plt
import scienceplots
plt.style.use("science")
```

```python
def euler(t0, y0, df, h, n):
    t = np.arange(t0, n+h, h)
    m = t.size
    y = np.zeros(m)
    y[0] = y0
    for i in range(1, m):
        y[i] = y[i-1] + h * df(t[i-1], y[i-1])
    return t, y
```

```python
def error(t, y):
    n = t.size
    truey = np.zeros(n)
    for i in range(n):
        truey[i] = np.exp(t[i])
    diff = y - truey
    return (diff*diff).mean()
```
